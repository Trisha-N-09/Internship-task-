```python
df = pd.read_csv('/content/marketing_campaign.csv', sep='\t')

print("Dataset shape:", df.shape)
print("\nFirst 5 rows:")
display(df.head())

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())
```

```python
print("Education values:")
print(df["Education"].value_counts())

print("\nMarital Status values:")
print(df["Marital_Status"].value_counts())
```

```python
print("Date column data type:")
print(df["Dt_Customer"].dtype)

print("\nSample date values:")
print(df["Dt_Customer"].head(10))
```

```python
categorical_columns = df.select_dtypes(include="object").columns

for column in categorical_columns:
    print(f"\n{column}:")
    print(df[column].value_counts())
```

```python
print(df.dtypes)
```

```python
df["Dt_Customer"] = pd.to_datetime(
    df["Dt_Customer"],
    format="%d-%m-%Y"
)

print("New data type:")
print(df["Dt_Customer"].dtype)

print("\nSample converted dates:")
print(df["Dt_Customer"].head())
```

```python
# Fill missing Income values with the median income
median_income = df["Income"].median()

df["Income"] = df["Income"].fillna(median_income)

print("Median income used:", median_income)
print("Missing Income values after cleaning:", df["Income"].isnull().sum())
```

```python
# Remove duplicate rows
before = len(df)

df = df.drop_duplicates()

after = len(df)

print("Rows before:", before)
print("Rows after:", after)
print("Duplicate rows removed:", before - after)
```

```python
# Standardize column names
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

print("Updated column names:")
print(df.columns.tolist())
```

```python
# Check for negative values in numeric columns
numeric_columns = df.select_dtypes(include="number").columns

negative_values = (df[numeric_columns] < 0).sum()

print("Negative values in each numeric column:")
print(negative_values[negative_values > 0])
```

```python
print("Missing values after cleaning:")
print(df.isnull().sum()[df.isnull().sum() > 0])
```

```python
# Check for duplicate rows
duplicates = df.duplicated().sum()

print("Number of duplicate rows:", duplicates)
```

```python
# Check data types of all columns
print(df.dtypes)
```

```python
# Check unique values in categorical columns
print("Education values:")
print(df['education'].unique())

print("\nMarital Status values:")
print(df['marital_status'].unique())
```

```python
print("Education:")
print(df['education'].value_counts())

print("\nMarital Status:")
print(df['marital_status'].value_counts())
```

```python
# Inspect unusual marital status records
unusual = df[df['marital_status'].isin(['Alone', 'Absurd', 'YOLO'])]

print(unusual[['id', 'year_birth', 'education', 'marital_status', 'income']])
```

```python
# Group rare/unusual marital status categories
df['marital_status'] = df['marital_status'].replace(
    ['Alone', 'Absurd', 'YOLO'],
    'Other'
)

# Check the result
print(df['marital_status'].value_counts())
```

```python
# Statistical summary of numerical columns
df.describe()
```

```python
# Check suspicious birth years
print("Customers born before 1920:")
print(df[df['year_birth'] < 1920][['id', 'year_birth', 'income']])

print("\nCustomers with very high income:")
print(df[df['income'] > 200000][['id', 'year_birth', 'income']])
```

```python
import numpy as np
```

```python
# Replace invalid birth years with NaN
df.loc[df['year_birth'] < 1920, 'year_birth'] = np.nan

# Replace suspicious income value with NaN
df.loc[df['income'] == 666666, 'income'] = np.nan

# Check the result
print("Missing values after handling anomalies:")
print(df[['year_birth', 'income']].isnull().sum())
```

```python
# Fill missing birth years with median
df['year_birth'] = df['year_birth'].fillna(df['year_birth'].median())

# Fill missing income with median
df['income'] = df['income'].fillna(df['income'].median())

# Check missing values again
print("Missing values after final cleaning:")
print(df.isnull().sum())
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
df['education'].value_counts().plot(kind='bar')
plt.title('Distribution of Customers by Education')
plt.xlabel('Education')
plt.ylabel('Number of Customers')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8,5))

df['marital_status'].value_counts().plot(kind='bar')

plt.title('Distribution of Customers by Marital Status')
plt.xlabel('Marital Status')
plt.ylabel('Number of Customers')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df['income'], bins=30)

plt.title('Distribution of Customer Income')
plt.xlabel('Income')
plt.ylabel('Number of Customers')

plt.show()
```

```python
# Calculate total spending for each product category

spending = {
    'Wines': df['mntwines'].sum(),
    'Fruits': df['mntfruits'].sum(),
    'Meat Products': df['mntmeatproducts'].sum(),
    'Fish Products': df['mntfishproducts'].sum(),
    'Sweet Products': df['mntsweetproducts'].sum(),
    'Gold Products': df['mntgoldprods'].sum()
}

plt.figure(figsize=(10,5))

plt.bar(spending.keys(), spending.values())

plt.title('Total Customer Spending by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Total Spending')

plt.xticks(rotation=45)
plt.show()
```

```python
# Calculate average purchases by channel

purchase_channels = {
    'Web': df['numwebpurchases'].mean(),
    'Catalog': df['numcatalogpurchases'].mean(),
    'Store': df['numstorepurchases'].mean()
}

plt.figure(figsize=(8,5))

plt.bar(purchase_channels.keys(), purchase_channels.values())

plt.title('Average Purchases by Channel')
plt.xlabel('Purchase Channel')
plt.ylabel('Average Number of Purchases')

plt.show()
```

```python
# Count campaign responses

response_counts = df['response'].value_counts()

plt.figure(figsize=(7,5))

plt.bar(['Did Not Respond', 'Responded'],
        [response_counts.get(0, 0), response_counts.get(1, 0)])

plt.title('Marketing Campaign Response')
plt.xlabel('Response')
plt.ylabel('Number of Customers')

plt.show()

print("Response counts:")
print(response_counts)

print("\nResponse rate:",
      round(df['response'].mean() * 100, 2), "%")
```

```python
# Compare average income by campaign response

income_response = df.groupby('response')['income'].mean()

print("Average income:")
print(income_response)

plt.figure(figsize=(7,5))

plt.bar(
    ['Did Not Respond', 'Responded'],
    [income_response[0], income_response[1]]
)

plt.title('Average Income by Campaign Response')
plt.xlabel('Campaign Response')
plt.ylabel('Average Income')

plt.show()
```

```python
# Count customers who accepted each campaign

campaigns = {
    'Campaign 1': df['acceptedcmp1'].sum(),
    'Campaign 2': df['acceptedcmp2'].sum(),
    'Campaign 3': df['acceptedcmp3'].sum(),
    'Campaign 4': df['acceptedcmp4'].sum(),
    'Campaign 5': df['acceptedcmp5'].sum()
}

print("Campaign acceptance counts:")
for campaign, count in campaigns.items():
    print(campaign, ":", count)

plt.figure(figsize=(8,5))

plt.bar(campaigns.keys(), campaigns.values())

plt.title('Marketing Campaign Performance')
plt.xlabel('Campaign')
plt.ylabel('Number of Customers Accepted')

plt.show()
```

```python
# Website visits vs web purchases

plt.figure(figsize=(8,5))

plt.scatter(
    df['numwebvisitsmonth'],
    df['numwebpurchases'],
    alpha=0.5
)

plt.title('Website Visits vs Web Purchases')
plt.xlabel('Number of Website Visits per Month')
plt.ylabel('Number of Web Purchases')

plt.show()
```

```python
# Select numerical columns
numeric_df = df.select_dtypes(include=['int64', 'float64'])

# Calculate correlation
correlation = numeric_df.corr()

# Display correlation with campaign response
print(correlation['response'].sort_values(ascending=False))
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation,
    cmap='coolwarm',
    center=0
)

plt.title('Correlation Heatmap')
plt.show()
```
