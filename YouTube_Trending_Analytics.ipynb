Analyze trending videos across regions to identify category, sentiment, engagement and trending-duration patterns.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('../data/youtube_trending_raw.csv')
print(df.shape)
df.head()
```

## Data Cleaning
Remove duplicate videos and fill missing comments using the median.

```python
df=df.drop_duplicates('Video_ID').copy()
df['Comments']=df['Comments'].fillna(df['Comments'].median())
df['Publish_Date']=pd.to_datetime(df['Publish_Date'])
print('Clean rows:',len(df),'Missing values:',df.isna().sum().sum())
```

## Category Analysis

```python
category=df.groupby('Category').agg(Videos=('Video_ID','count'),Avg_Views=('Views','mean'),Avg_Engagement=('Engagement_Rate','mean'),Avg_Sentiment=('Sentiment_Score','mean'),Avg_Trending_Days=('Trending_Days','mean')).reset_index()
category.round(2)
```

```python
plt.figure(figsize=(9,5)); plt.bar(category['Category'],category['Avg_Views']); plt.title('Average Views by Category'); plt.xticks(rotation=45,ha='right'); plt.ylabel('Average Views'); plt.tight_layout(); plt.show()
```

## Region-wise Comparison

```python
region=df.groupby('Region').agg(Avg_Views=('Views','mean'),Avg_Engagement=('Engagement_Rate','mean'),Avg_Sentiment=('Sentiment_Score','mean')).reset_index(); region.round(2)
```

## Sentiment and Trending Duration

```python
corr=df[['Trending_Days','Views']].corr().iloc[0,1]; print('Trending days vs views correlation:',round(corr,3))
```

## Conclusion
Category and region comparisons, sentiment scores, engagement rates and trending duration provide a data-storytelling view of what characteristics are associated with high visibility. Correlation is an association and does not prove causation.
