Predict customer lifetime value from purchase behavior using Frequency, Recency, Average Order Value (AOV), and tenure.

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('../data/customer_ltv_raw.csv')
print(df.shape)
df.head()
```

## Data preprocessing
Missing numerical values are filled using the median and duplicate customer IDs are removed.

```python
df['Average_Order_Value']=df['Average_Order_Value'].fillna(df['Average_Order_Value'].median())
df['Recency_Days']=df['Recency_Days'].fillna(df['Recency_Days'].median())
df=df.drop_duplicates('Customer_ID').copy()
print('Missing values:',df.isna().sum().sum())
```

## Feature engineering
The project brief identifies Frequency, Recency and AOV as core behavioral features. Tenure is included as an additional customer-history feature.

```python
features=['Frequency','Recency_Days','Average_Order_Value','Tenure_Months']
X=df[features]
y=df['Lifetime_Value']
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=.2,random_state=42)
```

## Random Forest regression model

```python
model=RandomForestRegressor(n_estimators=250,random_state=42,max_depth=12)
model.fit(X_train,y_train)
pred=model.predict(X_test)
mae=mean_absolute_error(y_test,pred)
rmse=np.sqrt(mean_squared_error(y_test,pred))
r2=r2_score(y_test,pred)
print('MAE:',round(mae,2))
print('RMSE:',round(rmse,2))
print('R2:',round(r2,3))
```

```python
plt.figure(figsize=(8,5))
plt.scatter(y_test,pred,alpha=.45)
plt.xlabel('Actual LTV'); plt.ylabel('Predicted LTV')
plt.title('Actual vs Predicted Customer LTV')
plt.show()
```

## Customer segmentation

```python
df['Predicted_LTV']=model.predict(df[features])
df['LTV_Segment']=pd.qcut(df['Predicted_LTV'],q=4,labels=['Low','Medium','High','Very High'])
df[['Customer_ID','Predicted_LTV','LTV_Segment']].head()
```

## Conclusion
The model converts customer purchase behavior into predicted lifetime value and segments customers into four relative LTV groups. These segments can support targeted marketing and customer-retention planning. Model metrics should be interpreted on the held-out test set.
