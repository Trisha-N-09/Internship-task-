# Retail Business Performance & Profitability Analysis\n## Objective\nAnalyze transactional retail data to uncover profit-draining categories, understand inventory turnover, and identify seasonal product behavior.\n

```python
import pandas as pd\nimport numpy as np\nimport matplotlib.pyplot as plt\n
```

```python
df = pd.read_csv('../data/retail_transactions_raw.csv')\nprint('Raw shape:', df.shape)\ndf.head()
```

## Data Cleaning\nRemove duplicate Order_ID records and fill missing discounts with 0.

```python
df = df.drop_duplicates(subset=['Order_ID']).copy()\ndf['Discount'] = df['Discount'].fillna(0)\ndf['Order_Date'] = pd.to_datetime(df['Order_Date'])\nprint('Clean shape:', df.shape)\nprint('Missing values:', df.isna().sum().sum())
```

## Category Analysis

```python
category = df.groupby('Category').agg(Sales=('Sales','sum'), Profit=('Profit','sum'), Avg_Inventory_Days=('Inventory_Days','mean')).reset_index()\ncategory['Profit_Margin_%'] = category['Profit']/category['Sales']*100\ncategory.round(2)
```

```python
plt.figure(figsize=(9,5))\nplt.bar(category['Category'], category['Profit'])\nplt.title('Profit by Category')\nplt.xlabel('Category'); plt.ylabel('Profit')\nplt.tight_layout()\nplt.show()
```

## Inventory vs Profitability

```python
corr = df[['Inventory_Days','Profit']].corr().iloc[0,1]\nprint('Correlation:', round(corr,3))\nplt.figure(figsize=(9,5))\nplt.scatter(df['Inventory_Days'], df['Profit'], alpha=0.35)\nplt.title(f'Inventory Days vs Profit (r={corr:.2f})')\nplt.xlabel('Inventory Days'); plt.ylabel('Profit')\nplt.tight_layout(); plt.show()
```

## Seasonal Behavior

```python
season = df.groupby('Season').agg(Sales=('Sales','sum'), Profit=('Profit','sum')).reset_index()\nseason['Profit_Margin_%'] = season['Profit']/season['Sales']*100\nseason.round(2)
```

## Conclusion\nUse category and sub-category profitability together with inventory days to identify products that need pricing, promotion, or inventory-control attention. Seasonal trends can guide stock planning and campaign timing.
